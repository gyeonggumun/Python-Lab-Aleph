```python
# input으로 level 받아 복합 조건 적용
ip = "203.0.113.5"
level = int(input("레벨? "))
if level >= 10 and ip not in ["127.0.0.1", "192.168.1.1"]:
    print(f"[HIGH] 외부 IP {ip} 심각 경보")
elif level >= 7:
    print(f"[MED] 레벨 {level} 주의")
else:
    print("[LOW] 정상")
```

```python
import json
alert = {"rule": 5712, "level": 10, "ip": "1.1.1.1"}
s = json.dumps(alert, ensure_ascii=False)
print(s)                          # {"rule": 5712, "level": 10, "ip": "1.1.1.1"}
d = json.loads(s)
print(d["level"], type(d))        # 10 <class 'dict'>
```

```python
def severity(level):
    if level >= 10: return "High"
    if level >= 7:  return "Medium"
    return "Low"

print(severity(10))  # High
print(severity(8))   # Medium
print(severity(3))   # Low
```

```python
def classify(alert):
    sev = severity(alert["level"])
    return {"ip": alert["ip"], "severity": sev}

alerts = [
    {"level": 12, "ip": "1.1.1.1"},
    {"level": 9,  "ip": "2.2.2.2"},
    {"level": 5,  "ip": "3.3.3.3"},
]
for a in alerts:
    print(classify(a))
```
