```python
# input으로 level 받아 복합 조건 적용
ip = "203.0.113.5"
level = int(input("레벨? "))
if level >= 10 and ip not in ["127.0.0.1", "192.168.1.1"]:
    print(f"[HIGH] 외부 IP {ip} 심각 경보")
elif level >= 7:
    print(f"[MED] 레벨 {level} 주의")
else:
    print("[LOW] 정상")
```
